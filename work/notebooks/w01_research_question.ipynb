## 3. Quick look at the data (2-3 real numbers)

The ML-01 hand rule defines stale and visible pages as those last updated at least 180 days ago with at least 500 impressions in the previous 90 days. I compared the share marked as declining in that group with the share in the full starter sample. The rule flagged only 17 pages. Of those, 94.1% were marked as declining, compared with 54.2% overall. That small group is a useful starting point for testing whether a broader queue can surface more pages worth reviewing.

```python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/vuchau0802/run-starter-notebooks/"
    "main/data/raw/content_refresh_anonymized.csv"
)
columns = ["days_since_last_update", "impressions_90d", "trend_direction"]
df = pd.read_csv(url, usecols=columns)

stale_and_visible = (
    df["days_since_last_update"].ge(180)
    & df["impressions_90d"].ge(500)
)
declining = df["trend_direction"].eq("down")

print(f"Pages flagged by the rule: {stale_and_visible.sum():,}")
print(f"Marked as declining among flagged pages: {declining[stale_and_visible].mean():.1%}")
print(f"Marked as declining overall: {declining.mean():.1%}")
```
